```python
# Q6
median_val_hp = df.horsepower.median()
most_frequent_val = df.horsepower.value_counts().index[0]
df.horsepower = df.horsepower.fillna(most_frequent_val)
new_med_val_hp = df.horsepower.median()
new_med_val_hp - median_val_hp
```

```python
# Q7
df_asian = df[df['origin'] == 'Asia']
df1 = df_asian[['vehicle_weight','model_year']]
df2 = df1.iloc[:7]
X = df2.to_numpy()
X_T = X.T

XTX = X_T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = XTX_inv.dot(X_T).dot(y)
w.sum().round(3)
```
